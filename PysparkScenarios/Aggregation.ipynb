```python
##### Imports ######
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
emp_data = [
(1,'manish',26,20000,'india','IT'),
(2,'rahul',None,40000,'germany','engineering'),
(3,'pawan',12,60000,'india','sales'),
(4,'roshini',44,None,'uk','engineering'),
(5,'raushan',35,70000,'india','sales'),
(6,None,29,200000,'uk','IT'),
(7,'adam',37,65000,'us','IT'),
(8,'chris',16,40000,'us','sales'),
(None,None,None,None,None,None),
(7,'adam',37,65000,'us','IT')
]

emp_schema = ['id','name','age','salary','country','dept']

emp_df = spark.createDataFrame(data=emp_data,schema=emp_schema) 
```

```python
emp_df.show()
```

```python
emp_df.count()
```

```python
emp_df.select(count("name")).show()
```

```python
emp_df.select(count("id")).show()
```

```python
emp_df.count()
```

```python
emp_df.select(count("country")).show()
```

```python
emp_df.select(count("*")).show()
```

```python
emp_df.select(sum("salary").alias("Total_Salary"),
              max("salary").alias("Maximum_Salary"),
              min("salary").alias("Minimum_Salary")).show()
```

```python
emp_df.select(sum("salary"),
              count("salary"),
              avg("salary").cast("int").alias("avg_salary")).show()
```
