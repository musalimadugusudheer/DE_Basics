```python
data=[(10 ,'Anil',50000, 18),
(11 ,'Vikas',75000,  16),
(12 ,'Nisha',40000,  18),
(13 ,'Nidhi',60000,  17),
(14 ,'Priya',80000,  18),
(15 ,'Mohit',45000,  18),
(16 ,'Rajesh',90000, 10),
(17 ,'Raman',55000, 16),
(18 ,'Sam',65000,   17)]


schema = ['id','name','salary','mgr_id']

manager_df = spark.createDataFrame(data=data, schema=schema)
```

```python
data1=[(19 ,'Sohan',50000, 18),
(20 ,'Sima',75000,  17)]


schema1 = ['id','name','salary','mgr_id']

manager_df1 = spark.createDataFrame(data=data1, schema=schema1)
```

```python
manager_df1.show()
```

```python
manager_df.union(manager_df1).show()
```

```python
manager_df.union(manager_df1).count()
```

```python
manager_df.unionAll(manager_df1).count()
```

```python
data2=[(10 ,'Anil',50000, 18),
(11 ,'Vikas',75000,  16),
(12 ,'Nisha',40000,  18),
(13 ,'Nidhi',60000,  17),
(14 ,'Priya',80000,  18),
(15 ,'Mohit',45000,  18),
(16 ,'Rajesh',90000, 10),
(17 ,'Raman',55000, 16),
(18 ,'Sam',65000,   17),
(18 ,'Sam',55000,   17),
(18 ,'Sam',65000,   17)]

schema2 = ['id','name','salary','mgr_id']

duplicate_manager_df = spark.createDataFrame(data=data2, schema=schema2)
```

```python
duplicate_manager_df.show()
```

```python
duplicate_manager_df.union(manager_df1).count()
```

```python
duplicate_manager_df.unionAll(manager_df1).count()
```

```python
manager_df1.createOrReplaceTempView("manager_df1_tbl")
duplicate_manager_df.createOrReplaceTempView("duplicate_manager_df_tbl")
```

```python
spark.sql("""
          select * from manager_df1_tbl 
          union
          select * from duplicate_manager_df_tbl
          """).count()
```

```python
spark.sql("""
          select * from manager_df1_tbl 
          union All
          select * from duplicate_manager_df_tbl
          """).count()
```

```python
spark.sql("""
          select * from duplicate_manager_df_tbl 
          union
          select * from duplicate_manager_df_tbl
          """).show()
```
